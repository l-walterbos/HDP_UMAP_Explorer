```python
novac = dcomb['element.B2'].dropna().index

fig, ax = plt.subplots()

ax.plot(dcomb['Icohp.B1.avg'],dcomb['bandgap'],'b.')
ax.plot(dcomb.loc[novac]['Icohp.B2.avg'],dcomb.loc[novac]['bandgap'],'b.')
ax.set_xlabel('ICOBI value')
ax.set_ylabel('bandgap (eV)')

fig.show()
```

```python
dcomb['specie.B2'] = dcomb['specie.B2'].fillna('Vac')
B1counts = dcomb.groupby('specie.B1').count()
B2counts = dcomb.groupby('specie.B2').count()
counts = B1counts.add(B2counts,fill_value=0)
# counts

counts = counts['bandgap']

B1gaps = dcomb.groupby('specie.B1')['bandgap'].sum()
B2gaps = dcomb.groupby('specie.B2')['bandgap'].sum()
Bgaps = B1gaps.add(B2gaps,fill_value=0)

B1icobi = dcomb.groupby('specie.B1')['charge.B1'].sum()
B2icobi = dcomb.groupby('specie.B2')['charge.B2'].sum()
Bicobi = B1icobi.add(B2icobi,fill_value=0)


elavg_gaps = Bgaps.div(counts,fill_value=0)
elavg_icobis = Bicobi.div(counts,fill_value=0)

print(elavg_icobis)

fig, ax = plt.subplots()
ax.plot(elavg_icobis, elavg_gaps,'r.')
ax.set_xlabel('specieal Avg. charge')
ax.set_ylabel('bandgap (eV)')
fig.show()
```
